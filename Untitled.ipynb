```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
import os

output_folder = "thesis_outputs"
os.makedirs(output_folder, exist_ok=True)



sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("switch_inventory (1).csv", sep=";")
```

```python
df["Utilization_pct"] = (
    df["Utilization_pct"]
    .str.replace("%", "", regex=False)   # Remove %
    .str.replace(" ", "", regex=False)   # Remove spaces
    .str.replace(",", ".", regex=False)  # Replace comma with decimal point
    .astype(float)                       # Convert to numbers
)
```

```python
df.head()
```

```python
#Analysis 1: Total unused ports per location
```

```python
Total_unused_edge_ports = (
df.groupby("Location")["unused_edge_ports"]
.sum()
.sort_values(ascending=False))
Total_unused_edge_ports
```

```python
#Visualization Total Available unused edge ports per location
```

```python
plt.figure(figsize=(12, 6))

colors = sns.color_palette("Set2", n_colors=len(Total_unused_edge_ports))

ax = Total_unused_edge_ports.plot(
    kind="bar",
    color=colors
)

# Add values above bars
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.0f",
        padding=3,
        fontsize=9
    )

plt.title(
    "Unused Edge Ports by Location",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Location")
plt.ylabel("Available Ports")
plt.xticks(rotation=45, ha="right")

plt.grid(
    axis="y",
    linestyle="--",
    alpha=0.3
)

plt.savefig(
    "unused_edge_ports.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Available unused edge ports per switch per floor per location
```

```python
unused_ports_table = (
    df[
        [
            "Location",
            "Floor",
            "Switch",
            "unused_edge_ports"
        ]
    ]
    .sort_values(
        ["Location", "Floor", "Switch"]
    )
    .reset_index(drop=True)
)

display(unused_ports_table)
```

```python
#Vsualisation for available unused edge ports per switch per loction per floor 
```

```python
sns.set_theme(style="whitegrid")

locations = df["Location"].unique()

for loc in locations:

    subset = df[df["Location"] == loc].copy()

    # Sort by floor and switch
    subset = subset.sort_values(["Floor", "Switch"])

    labels = (
    subset["Floor"].astype(str)
    + " | "
    + subset["Switch"].astype(int).astype(str)
)

    colors = []

    for x in subset["unused_edge_ports"]:
        if x < 5:
            colors.append("#d62728")      # red
        elif x < 10:
            colors.append("#ffbf00")      # yellow
        else:
            colors.append("#2ca02c")      # green

    plt.figure(figsize=(12, 8))

    bars = plt.barh(
        labels,
        subset["unused_edge_ports"],
        color=colors
    )

    plt.bar_label(
        bars,
        fmt='%d',
        padding=4,
        fontsize=10,
        fontweight='bold'
    )

    plt.title(
        f"Available Unused Edge Ports - {loc}",
        fontsize=16,
        fontweight="bold"
    )

    plt.xlabel("Unused Edge Ports", fontsize=12)
    plt.ylabel("Floor | Switch", fontsize=12)

    plt.grid(axis="x", linestyle="--", alpha=0.4)

    sns.despine()

    # Save a separate PNG for this location
    filename = f"unused_edge_ports_{loc}.png"

    plt.savefig(
        filename,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    plt.close()
```

```python
#Unused edge ports equivalent to 48 ports Swicth per location
```

```python
location_summary = (
    df.groupby("Location", as_index=False)
      .agg(
          Current_Switches=("Switch", "nunique"),
          Total_Available_Unused_Ports=("unused_edge_ports", "sum")
      )
)

location_summary["Switch_Equivalent_48p"] = (
    location_summary["Total_Available_Unused_Ports"] / 48
).round(2)

location_summary["Switches_Purchase_Delayed"] = (
    location_summary["Total_Available_Unused_Ports"] // 48
).astype(int)

location_summary["Remaining_Ports"] = (
    location_summary["Total_Available_Unused_Ports"] % 48
).astype(int)
display(location_summary)
```

```python
#Visulation of 48 port switch equivalent and swicth purchased delayed
```

```python

plot_data = location_summary.sort_values(
    "Switch_Equivalent_48p",
    ascending=True
)

plt.figure(figsize=(12, 8))

colors = sns.color_palette(
    "Set2",
    n_colors=len(plot_data)
)

bars = plt.barh(
    plot_data["Location"],
    plot_data["Switch_Equivalent_48p"],
    color=colors
)

# Add switch equivalent + potential delayed purchases
for i, bar in enumerate(bars):

    equivalent = plot_data.iloc[i]["Switch_Equivalent_48p"]
    delayed = plot_data.iloc[i]["Switches_Purchase_Delayed"]

    plt.text(
        bar.get_width() + 0.03,
        bar.get_y() + bar.get_height() / 2,
        f"{equivalent:.2f} ({delayed} switch{'es' if delayed != 1 else ''})",
        va="center",
        fontsize=10
    )

plt.xlabel(
    "48-port switch equivalent",
    fontsize=12
)

plt.ylabel(
    "Location",
    fontsize=12
)

plt.title(
    "Potential Switch Purchase Deferral (Unused Edge Ports)",
    fontsize=15,
    fontweight="bold"
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.tight_layout()

# Save the plot
filename = "potential_switch_deferral.png"

plt.savefig(
    filename,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

```

```python
#Cost analysis for unused edge ports
```

```python
SWITCH_COST_EUR = 1268

location_summary["Potential_Cost_Deferral_EUR"] = (
    location_summary["Switches_Purchase_Delayed"] * SWITCH_COST_EUR
)

display(location_summary)
```

```python
#Visulization for cost analysis for unused edge ports
```

```python
plot_data = location_summary.sort_values(
    "Potential_Cost_Deferral_EUR",
    ascending=True
)

plt.figure(figsize=(12, 8))

colors = sns.color_palette(
    "Set2",
    n_colors=len(plot_data)
)

bars = plt.barh(
    plot_data["Location"],
    plot_data["Potential_Cost_Deferral_EUR"],
    color=colors
)

for i, bar in enumerate(bars):

    cost = plot_data.iloc[i]["Potential_Cost_Deferral_EUR"]
    switches = plot_data.iloc[i]["Switches_Purchase_Delayed"]

    plt.text(
        cost + 30,
        bar.get_y() + bar.get_height() / 2,
        f"€{cost:,.0f} ({switches} switch{'es' if switches != 1 else ''})",
        va="center",
        fontsize=9
    )

plt.xlabel("Potential cost deferral (€)")
plt.ylabel("Location")

plt.title(
    "Potential Switch Purchase Cost Deferral by Location (Unused Edge ports)",
    fontsize=15,
    fontweight="bold"
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.tight_layout()

# Save the plot
filename = "potential_cost_deferral.png"

plt.savefig(
    filename,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

```

```python
#Total potential expansion capacity if baseline ports are reclaimed (per location and switch equivalent)
```

```python
location_expansion= (
    df.groupby("Location", as_index=False)
      .agg(
          Current_Switches=("Switch", "nunique"),
          Total_Expansion_Capacity=("Potential_Expansion_capacity", "sum")
      )
)

location_expansion["Switch_Equivalent_48p"] = (
    location_expansion["Total_Expansion_Capacity"] / 48
).round(2)

location_expansion["Switches_Purchase_Delayed"] = (
    location_expansion["Total_Expansion_Capacity"] // 48
).astype(int)

location_expansion["Remaining_Ports"] = (
    location_expansion["Total_Expansion_Capacity"] % 48
).astype(int)

display(location_expansion)
```

```python
#Visualisation for Total Expansion Capacity
```

```python
plot_data = location_expansion.sort_values(
    "Switch_Equivalent_48p",
    ascending=True
)

plt.figure(figsize=(12, 8))

colors = sns.color_palette(
    "Set2",
    n_colors=len(plot_data)
)

bars = plt.barh(
    plot_data["Location"],
    plot_data["Switch_Equivalent_48p"],
    color=colors
)

# Add switch equivalent + potential delayed purchases
for i, bar in enumerate(bars):

    equivalent = plot_data.iloc[i]["Switch_Equivalent_48p"]
    delayed = plot_data.iloc[i]["Switches_Purchase_Delayed"]

    plt.text(
        bar.get_width() + 0.03,
        bar.get_y() + bar.get_height() / 2,
        f"{equivalent:.2f} ({delayed} switch{'es' if delayed != 1 else ''})",
        va="center",
        fontsize=10
    )

plt.xlabel(
    "48-port switch equivalent",
    fontsize=12
)

plt.ylabel(
    "Location",
    fontsize=12
)

plt.title(
    "Total Potential Switch Purchase Deferral",
    fontsize=15,
    fontweight="bold"
)

plt.grid(
    axis="x",
    linestyle="--",
    alpha=0.3
)

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

# Save the plot
filename = "total_potential_switch_deferral.png"

plt.savefig(
    filename,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


```
